```python
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam rupiah): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").lower()

if jenis_kendaraan == "motor":
    level_risiko = 2  # sedang
else:
    level_risiko = 1  # rendah

if usia_pengemudi < 20 or usia_pengemudi > 60:
    level_risiko = 3  # tinggi

if area_penggunaan == "perkotaan":
    if level_risiko < 3:
        level_risiko += 1

if sistem_pengaman == "tidak":
    if level_risiko < 3:
        level_risiko += 1
elif sistem_pengaman == "ya":
    pass
else:
    pass

if level_risiko == 3:
    kategori_risiko = "tinggi"
elif level_risiko == 2:
    kategori_risiko = "sedang"
else:
    kategori_risiko = "rendah"

if kategori_risiko == "tinggi":
    persentase_premi = 0.1
elif kategori_risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi_asuransi = persentase_premi * nilai_kendaraan

pesan_doraemon = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if kategori_risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon"
)


print("\n=== HASIL ANALISIS ASURANSI ===")
print(f"1. Kategori risiko: {kategori_risiko}")
print(f"2. Premi asuransi yang perlu dibayarkan: Rp{premi_asuransi:,.2f}")
print(f"3. Pesan Doraemon: {pesan_doraemon}")
```
